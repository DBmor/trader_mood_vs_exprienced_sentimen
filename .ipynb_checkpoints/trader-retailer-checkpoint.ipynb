```python

import pandas as pd
import numpy as np

trader_mood = pd.read_csv("data/fear_greed_index - fear_greed_index.csv")
sentimen_log = pd.read_csv("data/historical_data - historical_data.csv")
trader_mood.columns = ['col_epoch', 'col_value', 'col_mood', 'col_date']

print(trader_mood.shape)
print(sentimen_log.shape)
```

```python
print(trader_mood.columns)
print(sentimen_log.columns)
```

```python
sentimen_log['match_date'] = pd.to_datetime(sentimen_log['Timestamp IST'], dayfirst=True).dt.date
trader_mood['match_date'] = pd.to_datetime(trader_mood['col_date']).dt.date
```

```python
combo_data = pd.merge(sentimen_log, trader_mood, on='match_date', how='inner')
combo_data.shape
```

```python
combo_data = combo_data.dropna(subset=['Closed PnL', 'col_mood'])
combo_data = combo_data[combo_data['Size Tokens'] > 0]
combo_data.shape
```

```python
combo_data['win_flag'] = (combo_data['Closed PnL'] > 0).astype(int)
```

```python
mood_summary = combo_data.groupby('col_mood').agg(
    trades_count=('Closed PnL', 'count'),
    avg_pnl=('Closed PnL', 'mean'),
    mid_pnl=('Closed PnL', 'median'),
    win_ratio=('win_flag', 'mean')
).reset_index()

mood_summary
```

```python

combo_data['trade_value_usd'] = combo_data['Size Tokens'] * combo_data['Execution Price']


combo_data['long_volume'] = np.where(combo_data['Direction'].str.upper() == 'BUY', combo_data['trade_value_usd'], 0)
combo_data['short_volume'] = np.where(combo_data['Direction'].str.upper() == 'SELL', combo_data['trade_value_usd'], 0)


combo_data['gross_profit'] = np.where(combo_data['Closed PnL'] > 0, combo_data['Closed PnL'], 0)
combo_data['gross_loss'] = np.where(combo_data['Closed PnL'] < 0, np.abs(combo_data['Closed PnL']), 0)
```

```python
regime_analysis = combo_data.groupby('col_mood').agg(
    total_trades=('Closed PnL', 'count'),
    median_pnl=('Closed PnL', 'median'),
    mean_pnl=('Closed PnL', 'mean'),
    avg_position_usd=('trade_value_usd', 'mean'),
    sum_long_vol=('long_volume', 'sum'),
    sum_short_vol=('short_volume', 'sum'),
    total_gains=('gross_profit', 'sum'),
    total_losses=('gross_loss', 'sum')
).reset_index()


regime_analysis['profit_factor'] = regime_analysis['total_gains'] / (regime_analysis['total_losses'] + 1e-6)


regime_analysis['long_ratio'] = regime_analysis['sum_long_vol'] / (regime_analysis['sum_long_vol'] + regime_analysis['sum_short_vol'] + 1e-6)

regime_analysis[['col_mood', 'total_trades', 'median_pnl', 'avg_position_usd', 'profit_factor', 'long_ratio']]
```

```python

extreme_regimes = combo_data[combo_data['col_mood'].isin(['Extreme Fear', 'Extreme Greed'])]


fear_vs_greed = extreme_regimes.groupby('col_mood').agg(
    volume_trades=('Closed PnL', 'count'),
    mean_pnl=('Closed PnL', 'mean'),
    median_pnl=('Closed PnL', 'median'),
    avg_size_usd=('trade_value_usd', 'mean'),
    long_ratio=('long_volume', 'mean'),
    gross_gains=('gross_profit', 'sum'),
    gross_losses=('gross_loss', 'sum')
).reset_index()


fear_vs_greed['profit_factor'] = fear_vs_greed['gross_gains'] / (fear_vs_greed['gross_losses'] + 1e-6)


fear_vs_greed[['col_mood', 'volume_trades', 'mean_pnl', 'median_pnl', 'avg_size_usd', 'long_ratio', 'profit_factor']]
```

```python

trade_count = combo_data.groupby('Account').size().reset_index(name='total_trades')


trade_count['trader_type'] = trade_count['total_trades'].apply(
    lambda x: 'Frequent' if x > 50 else 'Infrequent'
)


combo_data = pd.merge(combo_data, trade_count[['Account', 'trader_type']], on='Account')
```

```python

profit = combo_data.groupby('Account')['Closed PnL'].sum().reset_index()


profit['performance'] = profit['Closed PnL'].apply(
    lambda x: 'Profitable' if x > 0 else 'Loss Making'
)


combo_data = pd.merge(combo_data, profit[['Account', 'performance']], on='Account')
```
